```python
# Heapify - Builds Min Heap from List
# Time: O(n), Space: O(1)
# Heapq only makes a Min Heap

A = [-4, 3, 1, 0, 2, 5, 10, 8, 12, 9]

import heapq
heapq.heapify(A)

A
```

```python
# Heap Push (Insert element)
# Time: O(log n)

heapq.heappush(A, 4)

A
```

```python
# Heap Pop (Extract min)
# Time: O(log n)

minn = heapq.heappop(A)

A, minn
```

```python
# Heap Sort
# Time: O(n log n), Space: O(n)
# NOTE: O(1) Space is possible via swapping, but this is complex

def heapsort(arr):
  heapq.heapify(arr)
  print("heapified:", arr)
  n = len(arr)
  new_list = [0] * n

  for i in range(n):
    minn = heapq.heappop(arr)
    new_list[i] = minn

  return new_list

heapsort([1, 3, 5, 7, 9, 2, 4, 6, 8, 0])
```

```python
# Heap Push Pop: Time: O(log n)
# First push new element to heap, then pop the smallest element from heap and return it

heapq.heappushpop(A, 99)
A
```

```python
# Max Heap
# Heapq only makes a Min Heap, so we can use negative values to simulate a Max Heap
B = [-4, 3, 1, 0, 2, 5, 10, 8, 12, 9]
n = len(B)

for i in range(n):
  B[i] = -B[i]

heapq.heapify(B)

print(B)
largest = -heapq.heappop(B)
print(largest)
```

So we basically forced the Min heap to function as a Max Heap instead.

```python
heapq.heappush(B, -7) # Insert 7 into max heap

B
```

```python
# Putting tuples of items on the heap

D = [5, 4, 3, 5, 4, 3, 5, 5, 4]

from collections import Counter

counter = Counter(D)

counter
```

```python
heap = []

for k, v in counter.items():
  heapq.heappush(heap, (v, k))

heap
```
