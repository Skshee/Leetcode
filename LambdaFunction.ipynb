Lambda function = A small anonymous function for a one time use (throw away function)
- They take any number of arguments, but have only 1 expression
Helps keep the namespace clean and is useful with higher-order functions
'sort()', 'map()', 'filter()', 'reduce()'
lambda parameters: expression

```python
double = lambda x : x**2
add = lambda x,y : x + y
max_value = lambda x,y : x if x > y else y
isEven = lambda x : True if x % 2 == 0 else False
print(double(4))
print(add(5,10))
print(max_value(5,10))
print(isEven(5))


```

Lambda Function with Built-in Functions

```python
# Using lambda with Using Lambda with sorted()
# Example - Sorting strings by length
words = ["apple", "pie", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)
```

```python
# Using Lambda with map()
# Map function - applies a given function to each item of an iterable (like list) and returns a list of the results.
numbers = [1, 2, 3, 4, 5]
doubled = list(map(lambda x: x * 2, numbers))
print(doubled)
```

```python
# Sorting dictionary by values - V.IMP USE CASE
dic = [
    {"name": "Harry", "age": 20},
    {"name": "Robin", "age": 21},
    {"name": "Kevin", "age": 19}
]

print("Sorted by age:")
print(sorted(dic, key=lambda x: x['age']))

dic2 = {
    "Harry": 20,
    "Robin": 21,
    "Kevin": 19}
print(sorted(dic2.items(), key=lambda x: x[1]))  # Sort by values
```
